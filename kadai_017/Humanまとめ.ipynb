```python
class Human:
  #属性
  def __init__(self, name, age):
    self.name = name
    self.age = age
  #メソッド
  def check_adult(self):
    if self.age >= 20:
      print("大人")
    else:
      print("大人でない")

human1 = Human("taro", 21)
human2 = Human("jiro", 19)
human3 = Human("saburo", 18)
human1.check_adult()
human2.check_adult()

#リスト
list = [human1 , human2, human3]

for hu in list:
  hu.check_adult()


```
